```python
# Drop high-missing columns
health_df = health_df.drop(columns=['slope', 'ca', 'thal'])
```

```python
#impute continuous values with median
for col in ['trestbps', 'chol', 'thalch', 'oldpeak']:
    health_df[col] = health_df[col].fillna(health_df[col].median())
```

```python
#impute categorical values with mode
for col in ['fbs', 'restecg', 'exang']:
    health_df[col] = health_df[col].fillna(health_df[col].mode()[0])
```
