```python
# Drop high-missing columns
health_df = health_df.drop(columns=['slope', 'ca', 'thal'])
```

```python
#impute continuous values with median
for col in ['trestbps', 'chol', 'thalch', 'oldpeak']:
    health_df[col] = health_df[col].fillna(health_df[col].median())
```

```python
#impute categorical values with mode
for col in ['fbs', 'restecg', 'exang']:
    health_df[col] = health_df[col].fillna(health_df[col].mode()[0])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Convert num into binary (0 = no disease, 1+ = disease)
health_df['heart_disease'] = (health_df['num'] > 0).astype(int)
```

```python
plt.figure(figsize=(7,5))
sns.barplot(data=health_df, x='heart_disease', y='age')
plt.xticks([0, 1], ['No Disease', 'Disease'])
plt.title("Average Age for Heart Disease vs No Disease")
plt.xlabel("Heart Disease")
plt.ylabel("Average Age")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=health_df, x='age', hue='heart_disease', kde=True, bins=20)
plt.title("Age Distribution: Heart Disease vs No Disease")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
health_df['age_group'] = pd.cut(
    health_df['age'],
    bins=[20, 30, 40, 50, 60, 70, 80],
    labels=['20-29', '30-39', '40-49', '50-59', '60-69', '70-79']
)
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=health_df, x='age_group', y='heart_disease')
plt.title("Heart Disease Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Heart Disease Rate")
plt.show()
```
