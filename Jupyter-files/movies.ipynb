```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.scatterplot(data=movies_df, x='Budget', y='Revenue')
```

```python
# Scatter plots
plt.figure(figsize=(10,5))
plt.scatter(movies_df['DirectorsRating'], movies_df['Rating'], alpha=0.5, label='Directors vs Rating')
plt.scatter(movies_df['WritersRating'], movies_df['Rating'], alpha=0.5, label='Writers vs Rating', color='orange')
plt.xlabel('Directors/Writers Rating')
plt.ylabel('Audience Rating')
plt.title('Audience Rating vs Critics Ratings')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(movies_df['Budget'], movies_df['Revenue'], alpha=0.5)
plt.xlabel('Budget')
plt.ylabel('Revenue')
plt.title('Revenue vs Budget')
plt.show()
```

```python
# Group by year and calculate average budget
budget_by_year = movies_df.groupby('Date')['Budget'].mean()

# Plot
plt.figure(figsize=(12,6))
plt.plot(budget_by_year.index, budget_by_year.values, marker='o', color='purple')
plt.xlabel('Year')
plt.ylabel('Average Budget')
plt.title('Average Movie Budget by Year')
plt.grid(True)
plt.show()
```

```python
year_counts = movies_df['Date'].value_counts().sort_index()
plt.figure(figsize=(12,6))
plt.plot(year_counts.index, year_counts.values)
plt.xlabel('Year')
plt.ylabel('Number of Movies')
plt.title('Movies Released per Year')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(movies_df['Rating'].dropna(), bins=20, color='skyblue', edgecolor='black')
plt.xlabel('Rating')
plt.ylabel('Number of Movies')
plt.title('Distribution of Audience Ratings')
plt.show()
```

```python
top_directors = movies_df.groupby('Director')['Rating'].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,6))
plt.barh(top_directors.index, top_directors.values, color='green')
plt.xlabel('Average Rating')
plt.title('Top 10 Directors by Average Audience Rating')
plt.gca().invert_yaxis()
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.scatter(movies_df['Budget'], movies_df['Rating'], alpha=0.6, color='green')
plt.xlabel('Budget')
plt.ylabel('Rating')
plt.title('Movie Budget vs Rating')
plt.grid(True)
plt.show()
```
